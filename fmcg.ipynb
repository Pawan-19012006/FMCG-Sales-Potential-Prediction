```python
df = df.drop('Item_Outlet_Sales', axis=1)
```

```python
df.head()
```

```python
X = df.drop('Item_Outlet_Sales_Log',axis=1)
y = df['Item_Outlet_Sales_Log']
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30,random_state=42)
```

```python
from xgboost import XGBRegressor

xgb = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)
xgb.fit(X_train,y_train)
y_pred = xgb.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error,r2_score

mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)
print(mae)
print(mse)
print(r2)
```

```python
from sklearn.linear_model import LinearRegression

lin = LinearRegression()
lin.fit(X_train,y_train)
y_pred_lin = lin.predict(X_test)
```

```python
mae = mean_absolute_error(y_test,y_pred_lin)
mse = mean_squared_error(y_test,y_pred_lin)
r2 = r2_score(y_test,y_pred_lin)
print(mae)
print(mse)
print(r2)
```

```python
from sklearn.ensemble import RandomForestRegressor

forest = RandomForestRegressor(random_state=42)
forest.fit(X_train,y_train)
y_pred_forest = forest.predict(X_test)
```

```python
mae = mean_absolute_error(y_test,y_pred_forest)
mse = mean_squared_error(y_test,y_pred_forest)
r2 = r2_score(y_test,y_pred_forest)
print(mae)
print(mse)
print(r2)
```

```python
importance = pd.DataFrame({
    'Feature' : X.columns,
    'Importance' : xgb.feature_importances_
})

importance = importance.sort_values(
    by='Importance',
    ascending=False
)

print(importance.head(20))
```

```python
importance = importance.sort_values(
    by='Importance',
    ascending=True
)

plt.figure(figsize=(10,8))

plt.barh(
    importance['Feature'],
    importance['Importance']
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("XGBoost Feature Importance")

plt.show()
```

From the feature importance analysis, we can conclude that the most influential factor affecting outlet sales is the type of the outlet. In particular, grocery-store-related outlet types show a significantly high impact on sales prediction. Another major factor influencing sales is the product price (MRP), indicating that pricing strategy plays an important role in retail performance.

Interestingly, outlet age also contributes to sales prediction. Older outlets may have stronger customer loyalty, better market presence, and more established purchasing patterns compared to newly established outlets. This suggests that long-standing outlets tend to generate more stable and predictable sales behavior.
